```python
# Cell 7: Convert ligand PDB to GROMACS coordinate file (.gro)
!gmx editconf -f ber-clean_gmx.pdb -o ber.gro
```

```python
# Cell 8: Merge protein and ligand coordinates into complex.gro
with open("nsp2_processed.gro", "r") as f:
    protein_lines = f.readlines()

with open("ber.gro", "r") as f:
    ligand_lines = f.readlines()

# Extract just the coordinate lines from the ligand (skipping title, count, and box vectors)
ligand_atoms = ligand_lines[2:-1]
ligand_atom_count = len(ligand_atoms)

# Update the total atom count line for the complex
current_protein_count = int(protein_lines[1].strip())
new_total_count = current_protein_count + ligand_atom_count
protein_lines[1] = f"{new_total_count}\n"

# The last line of the protein file contains the box vectors.
# We insert the ligand atoms right before those vectors, and carry over the box dimensions.
complex_lines = protein_lines[:-1] + ligand_atoms + [protein_lines[-1]]

with open("complex.gro", "w") as f:
    f.writelines(complex_lines)

print(f"Success! Merged {ligand_atom_count} ligand atoms with {current_protein_count} protein atoms.")
print(f"New system total: {new_total_count} atoms written to complex.gro")
```

```python
# Cell 9: View the end of the topology file
!tail -n 20 topol.top
```

```python
# Cell 10: Update topol.top with ligand parameters and molecule count
with open("topol.top", "r") as f:
    lines = f.readlines()

new_lines = []
inserted_prm = False
inserted_itp = False

for line in lines:
    new_lines.append(line)

    # 1. Insert ber.prm right after the main forcefield include line
    if "forcefield.itp" in line and not inserted_prm:
        new_lines.append('\n; Include ligand parameters\n#include "ber.prm"\n')
        inserted_prm = True

    # 2. Insert ber.itp right before the [ system ] directive
    if "[ system ]" in line and not inserted_itp:
        # Step back one position to insert it cleanly above the header
        new_lines.insert(-1, '; Include ligand topology\n#include "ber.itp"\n\n')
        inserted_itp = True

# 3. Append the ligand to the very bottom of the molecules list
new_lines.append("BER                 1\n")

with open("topol.top", "w") as f:
    f.writelines(new_lines)

print("Success! topol.top has been updated with ligand parameters and definitions.")
```

```python
# Cell 11: Define the simulation box dimensions using a dodecahedron
!gmx editconf -f complex.gro -o newbox.gro -bt dodecahedron -d 1.0 -c
```

```python
# Cell 12: Solvate the system with water
!gmx solvate -cp newbox.gro -cs spc216.gro -o solv.gro -p topol.top
```

```python
# Cell 13: Create the ions.mdp file required for genion
mdp_content = """
; ions.mdp - used as input into grompp to generate ions.tpr
; Parameters describing what to do during the simulation
integrator  = steep     ; Algorithm (steepest descent minimization)
emtol       = 1000.0    ; Stop minimization when the maximum force < 1000.0 kJ/mol/nm
emstep      = 0.01      ; Minimization step size
nsteps      = 50000     ; Maximum number of (minimization) steps to perform

; Parameters describing how to find the neighbors of each atom and how to calculate the interactions
nstlist         = 1         ; Frequency to update the neighbor list and long range forces
cutoff-scheme   = Verlet    ; Buffered neighbor searching
ns_type         = grid      ; Method to determine neighbor list (local cells)
coulombtype     = PME       ; Treatment of long range electrostatic interactions
rcoulomb        = 1.2       ; Short-range electrostatic cut-off
rvdw            = 1.2       ; Short-range Van der Waals cut-off
pbc             = xyz       ; Periodic Boundary Conditions in all directions
"""

with open("ions.mdp", "w") as f:
    f.write(mdp_content.strip())

print("Success! ions.mdp file has been written to your workspace.")
```

```python
# Cell 14: Assemble the system binary (.tpr) for adding ions
!gmx grompp -f ions.mdp -c solv.gro -p topol.top -o ions.tpr
```

```python
# Cell 16: Fix missing molecule name in ber.itp
with open("ber.itp", "r") as f:
    lines = f.readlines()

# Line 5 (index 4) is the blank line under the header where the name should be
lines[4] = "BER\t\t\t 3\n"

with open("ber.itp", "w") as f:
    f.writelines(lines)

print("Success! 'BER' molecule name added to ber.itp.")
```
